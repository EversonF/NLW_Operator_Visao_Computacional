# Reconhecimento de Imagem com Gemini API
Este notebook utiliza a API do Google Gemini para analisar uma imagem guardada na pasta `imagens`.

```python
import os
from google import genai
from dotenv import load_dotenv
from PIL import Image
from IPython.display import display
from pydantic import BaseModel, Field

# Carregar as variáveis de ambiente do arquivo .env
# (Lembre-se de adicionar sua API_KEY no arquivo .env antes de executar)
load_dotenv()

# Obter a chave da API do ambiente
api_key = os.getenv("API_KEY")
if not api_key:
    raise ValueError("A variável API_KEY não foi encontrada. Verifique seu arquivo .env.")

# Configurar o client do Gemini com a chave da API
client = genai.Client(api_key=api_key)
```

```python
# O modelo será especificado na chamada da API
# (recomendado: gemini-2.5-flash para multimodais)

# Definir o caminho para a imagem (exemplo com bird.jpg)
image_path = 'imagens/bird.jpg'

# Carregar e exibir a imagem para visualização
img = Image.open(image_path)

# Redimensionar a imagem apenas para visualização no notebook
img_display = img.copy()
img_display.thumbnail((400, 400))
display(img_display)
```

```python
# Definir a estrutura de saída usando Pydantic
class AnaliseImagem(BaseModel):
    descricao_geral: str = Field(description="Descrição detalhada do conteúdo da imagem em português")
    objetos_principais: list[str] = Field(description="Lista dos principais objetos ou elementos detectados na imagem")
    cores_predominantes: list[str] = Field(description="Lista das cores que mais se destacam na imagem")
    ambiente_ou_contexto: str = Field(description="O ambiente onde a foto foi tirada ou contexto (ex: natureza, cidade, interno)")

# Enviar a imagem e o prompt para a API requisitando structured output
prompt = "Analise esta imagem em detalhes e extraia as informações estruturadas conforme solicitado."
print("Analisando a imagem guardada...")

# Chamar a API passando o config com response_mime_type e response_schema
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=[prompt, img],
    config={
        'response_mime_type': 'application/json',
        'response_schema': AnaliseImagem,
    }
)

# Exibir a resposta que já virá no formato JSON conforme o schema
print("\nSaída Estruturada do Gemini:")
print("-" * 40)
print(response.text)
```
